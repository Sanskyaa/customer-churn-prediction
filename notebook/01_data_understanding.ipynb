```python
def plot_churn_by_category(data, column):
    plt.figure(figsize=(8, 5))
    
    sns.countplot(
        data=data,
        x=column,
        hue="Churn"
    )
    
    plt.title(f"{column} vs Customer Churn")
    plt.xlabel(column)
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

```python
plot_churn_by_category(df, "InternetService")
plot_churn_by_category(df, "PaymentMethod")
plot_churn_by_category(df, "TechSupport")
plot_churn_by_category(df, "OnlineSecurity")
```

```python
df["AvgMonthlySpend"] = np.where(
    df["tenure"] > 0,
    df["TotalCharges"] / df["tenure"],
    df["MonthlyCharges"]
)
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, np.inf],
    labels=["New", "Growing", "Mature", "Loyal"]
)
```

```python
df["TenureGroup"].value_counts()
```

```python
service_columns = [
    "PhoneService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]
```

```python
df["NumServices"] = (
    df[service_columns]
    .apply(lambda row: (row == "Yes").sum(), axis=1)
)
```

```python
df["NumServices"].value_counts().sort_index()
```

```python
high_value_threshold = df["MonthlyCharges"].median()

df["IsHighValue"] = (
    df["MonthlyCharges"] > high_value_threshold
).astype(int)
```

```python
df["IsHighValue"].value_counts()
```

```python
protection_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport"
]

df["HasProtection"] = (
    df[protection_columns]
    .eq("Yes")
    .any(axis=1)
    .astype(int)
)
```

```python
df["HasProtection"].value_counts()
```
